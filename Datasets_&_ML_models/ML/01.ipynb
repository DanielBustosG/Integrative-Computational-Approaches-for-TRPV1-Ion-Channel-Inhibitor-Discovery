# Deleting duplicates; IC50 based on median

```python
import statistics
```

```python
#Manejo de duplicados; para las moléculas que están más de una vez se considera un elemento dejando el valor de IC50 en base a la mediana

duplicated = df3[df3.molecule_chembl_id.duplicated()]
duplicated
for i in duplicated.molecule_chembl_id:
    median = statistics.median(pd.to_numeric(df3.standard_value[df3.molecule_chembl_id == i], downcast="float"))
    df3.loc[df3.molecule_chembl_id == i, 'standard_value'] = median
```

## **Data pre-processing of the bioactivity data**

### **Labeling compounds as either being active, inactive or intermediate**
The bioactivity data is in the IC50 unit. Compounds having values of less than 1000 nM will be considered to be **active** while those greater than 10,000 nM will be considered to be **inactive**. As for those values in between 1,000 and 10,000 nM will be referred to as **intermediate**. 

```python
bioactivity_class = []
for i in df3.standard_value:
  if float(i) >= 50000:
    bioactivity_class.append("inactive")
  elif float(i) <= 10000:
    bioactivity_class.append("active")
  else:
    bioactivity_class.append("intermediate")
```

### **Iterate the *molecule_chembl_id* to a list**

```python
mol_cid = []
for i in df3.molecule_chembl_id:
  mol_cid.append(i)
```

### **Iterate *canonical_smiles* to a list**

```python
canonical_smiles = []
for i in df3.canonical_smiles:
  canonical_smiles.append(i)
```

### **Iterate *standard_value* to a list**

```python
standard_value = []
for i in df3.standard_value:
  standard_value.append(i)
```

### **Combine the 4 lists into a dataframe**

```python
data_tuples = list(zip(mol_cid, canonical_smiles, bioactivity_class, standard_value))
df4 = pd.DataFrame( data_tuples,  columns=['molecule_chembl_id', 'canonical_smiles', 'bioactivity_class', 'standard_value'])
```
